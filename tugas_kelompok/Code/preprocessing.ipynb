```python
cap = df_clean["resolution_time_hours"].quantile(0.995)
df_clean["y"] = df_clean["resolution_time_hours"].clip(upper=cap)

cap
```

```python
cd = df_clean["Created Date"]
df_clean["created_year"] = cd.dt.year
df_clean["created_month"] = cd.dt.month
df_clean["created_day"] = cd.dt.day
df_clean["created_dayofweek"] = cd.dt.dayofweek
df_clean["created_hour"] = cd.dt.hour
df_clean["is_weekend"] = (df_clean["created_dayofweek"] >= 5).astype(int)
```

```python
drop_cols = [
    "Complaint Type", "Descriptor",
    "Created Date", "Closed Date",
    "Due Date", "Resolution Action Updated Date"
]
df_clean = df_clean.drop(columns=drop_cols, errors="ignore")
```

```python
cap = df_clean["resolution_time_hours"].quantile(0.995)
df_clean["y"] = df_clean["resolution_time_hours"].clip(upper=cap)
cap
```

```python
feature_cols = [
    "created_year", "created_month", "created_dayofweek", "created_hour", "is_weekend",
    "Borough", "Community Board", "Incident Zip", "City",
    "Location Type", "Address Type",
    "Latitude", "Longitude"
]

X = df_clean[feature_cols].copy()
y = df_clean["y"].copy()

X.shape, y.shape
```

```python
X["Incident Zip"] = X["Incident Zip"].astype(str)
X["Incident Zip"] = X["Incident Zip"].str.replace(r"\.0$", "", regex=True)
X["Incident Zip"] = X["Incident Zip"].replace(["nan", "NaN", "<NA>"], "Unknown")
X["Incident Zip"] = X["Incident Zip"].fillna("Unknown")
```

```python
def topk_other(series, k=30, other_label="Other"):
    series = series.fillna("Unknown").astype(str)
    top = series.value_counts().head(k).index
    return series.where(series.isin(top), other_label)

X["Incident Zip"] = topk_other(X["Incident Zip"], k=30)
X["Community Board"] = topk_other(X["Community Board"], k=30)
X["City"] = topk_other(X["City"], k=30)
```

```python
(X.isna().mean()*100).sort_values(ascending=False).head(15)
```

```python
df_model = X.copy()
df_model["y"] = y
df_model.to_csv("modeling_ready.csv", index=False)
```
